## 📦 1. Importar librerías

Primero importamos todas las herramientas que vamos a usar.
- **pandas**: para organizar datos en tablas
- **matplotlib**: para hacer gráficas
- **numpy**: para cálculos matemáticos

```python
# Importamos las librerías principales
import pandas as pd       # Manejo de datos en tablas
import matplotlib.pyplot as plt  # Visualizaciones
import numpy as np        # Cálculos numéricos

# Configuración para que las gráficas se vean bien
plt.rcParams['figure.figsize'] = (10, 6)  # Tamaño por defecto
plt.rcParams['font.family'] = 'DejaVu Sans'
plt.style.use('seaborn-v0_8-whitegrid')  # Estilo limpio

print('✅ Librerías importadas correctamente')
```

---
## 💰 2. Datos Económicos del Proyecto

Aquí cargamos los datos básicos del proyecto para hacer los primeros cálculos.

**📚 Fuentes:**
- Inversión total y 2026: Gobierno de Puebla / Ley de Egresos 2026 — `puebla.gob.mx`
- Infraestructura (km, líneas, estaciones, torres): Comunicado oficial del Estado
- Usuarios y habitantes beneficiados: Declaraciones oficiales del gobernador Armenta
- ⚠️ Los datos de habitantes (132,306 y 1.6M) son cifras oficiales **sin metodología pública disponible**

```python
# -----------------------------------------------
# DATOS GENERALES DEL PROYECTO
# Fuente: Gobierno de Puebla (oficial)
# -----------------------------------------------

datos_proyecto = {
    'inversion_total_mdp': 6752,        # Millones de pesos
    'inversion_2026_mdp': 1515,         # Solo el año 2026
    'extension_km': 13.6,
    'num_lineas': 4,
    'num_estaciones': 9,
    'num_torres': 90,
    'tarifa_pesos': 11,                 # Promedio 10-12 pesos
    'usuarios_diarios': 90000,
    'habitantes_directos': 132306,
    'habitantes_indirectos': 1600000,
    'colonias_beneficiadas': 45
}

# Convertimos a DataFrame para verlos ordenados
df_proyecto = pd.DataFrame(list(datos_proyecto.items()), 
                            columns=['Variable', 'Valor'])

print('📊 Datos generales del Cablebús de Puebla:')
print(df_proyecto.to_string(index=False))
```

---
## 🧮 3. Cálculo: Costo por Habitante

Este es uno de los indicadores más importantes. Nos dice cuánto dinero público se gasta por cada persona beneficiada.

```python
# -----------------------------------------------
# CÁLCULO: COSTO POR HABITANTE BENEFICIADO
# -----------------------------------------------

inversion_pesos = datos_proyecto['inversion_total_mdp'] * 1_000_000  # Convertir a pesos

# Calculamos el costo para cada escenario
costo_directo = inversion_pesos / datos_proyecto['habitantes_directos']
costo_indirecto = inversion_pesos / datos_proyecto['habitantes_indirectos']

print('💰 COSTO POR HABITANTE BENEFICIADO')
print('=' * 40)
print(f'Inversión total:           ${inversion_pesos:,.0f} pesos')
print(f'Habitantes directos:       {datos_proyecto["habitantes_directos"]:,}')
print(f'Habitantes indirectos:     {datos_proyecto["habitantes_indirectos"]:,}')
print()
print(f'Costo por beneficio DIRECTO:    ${costo_directo:,.0f} pesos/persona')
print(f'Costo por beneficio INDIRECTO:  ${costo_indirecto:,.0f} pesos/persona')
print()
print('📌 Contexto: El salario mínimo mensual en México (2026) es ~$3,248 pesos')
print(f'   El costo por habitante directo equivale a {costo_directo/3248:.1f} salarios mínimos mensuales')
```

---
## 🌳 4. Análisis Ambiental — El Caso de los Árboles

Uno de los puntos más controversiales: la cifra oficial de árboles afectados **cambió 4 veces** en solo 3 meses.

### 📚 Fuentes documentadas
| Fecha | Cifra | Fuente | URL |
|-------|-------|--------|-----|
| Enero 2026 | 980 | Secretaría de Infraestructura de Puebla (declaración inicial) | puebla.gob.mx |
| Febrero 2026 | 746 | Declaración de García Parra, vocero del gobierno | Angulo7 / LadoB |
| Marzo 2026 (inicio) | 350 | Ajuste técnico por reubicación de torres | El Sol de Puebla |
| Marzo 2026 (9 mar) | 100–116 | Nueva declaración oficial (García Parra) | Angulo7 (09/03/2026) |

> ⚠️ **Nota metodológica:** Estos datos provienen de declaraciones públicas rastreadas en medios de comunicación,
> no de un documento técnico oficial publicado. Se recomienda solicitar vía transparencia el dictamen técnico
> de impacto ambiental (MIA) para validar las cifras.

```python
# -----------------------------------------------
# DATOS: EVOLUCIÓN DE LA CIFRA OFICIAL DE ÁRBOLES
# ESTE ES EL CORAZÓN DE LA HIPÓTESIS
# -----------------------------------------------

# Creamos un DataFrame con la evolución de las cifras oficiales
# Fecha, fuente y cifra que dijo el gobierno en ese momento
df_arboles = pd.DataFrame({
    'Fecha': ['Ene 2026', 'Feb 2026', 'Mar 2026 (inicio)', 'Mar 2026 (9 mar)'],
    'Fuente': [
        'Secretaría de Infraestructura (inicial)',
        'Gobierno (García Parra) — Angulo7',
        'Ajuste técnico (torres reubicadas) — El Sol de Puebla',
        'Nueva declaración García Parra — Angulo7 09/03/2026'
    ],
    'Arboles_afectados': [980, 746, 350, 116],
    'Postura': [
        'Remoción y compensación con 10,000 nuevos',
        'No se talarán, solo se trasplantan',
        'Se modificó la ruta de algunas torres',
        'Solo 116 árboles serán removidos'
    ]
})

cifra_inicial = 980
cifra_final = 116

print('🌳 EVOLUCIÓN DE CIFRAS OFICIALES DE ÁRBOLES AFECTADOS')
print('=' * 60)
print(df_arboles.to_string(index=False))
print()
print(f'📉 Reducción total: {cifra_inicial - cifra_final} árboles "desaparecieron" del conteo oficial')
print(f'   Eso es una reducción del {((cifra_inicial - cifra_final)/cifra_inicial)*100:.1f}% en solo 3 meses')
print()
print('⚠️  LIMITACIÓN: Datos de declaraciones públicas en medios.')
print('    Se recomienda solicitar el dictamen técnico ambiental (MIA) vía transparencia.')
```

---
## 📈 5. Visualización — Cambio en la Cifra Oficial de Árboles

Vamos a graficar cómo cambió la cifra oficial. Una imagen vale más que mil palabras.

```python
# -----------------------------------------------
# GRÁFICA: EVOLUCIÓN DE LA CIFRA OFICIAL (4 puntos)
# -----------------------------------------------

import os

fig, ax = plt.subplots(figsize=(12, 6))

# Datos para la gráfica — ahora con 4 fechas
fechas = ['Enero\n2026', 'Febrero\n2026', 'Marzo 2026\n(inicio)', 'Marzo 2026\n(9 mar)']
cifras = [980, 746, 350, 116]
colores = ['#e74c3c', '#e67e22', '#f1c40f', '#27ae60']  # Rojo → Naranja → Amarillo → Verde

# Barras
bars = ax.bar(fechas, cifras, color=colores, width=0.5, edgecolor='white', linewidth=1.5)

# Agregar el número encima de cada barra
for bar, cifra in zip(bars, cifras):
    ax.text(bar.get_x() + bar.get_width()/2,
            bar.get_height() + 15,
            f'{cifra:,} árboles',
            ha='center', va='bottom',
            fontsize=12, fontweight='bold')

# Etiquetas y título
ax.set_title('¿Cuántos árboles afectará el Cablebús?\nLa cifra oficial cambió 4 veces en 3 meses (-88%)',
             fontsize=14, fontweight='bold', pad=15)
ax.set_ylabel('Número de árboles afectados', fontsize=12)
ax.set_ylim(0, 1200)

# Nota al pie con fuentes
ax.text(0.5, -0.18,
        'Fuentes: Secretaría de Infraestructura de Puebla, Angulo7 (09/03/2026), El Sol de Puebla | @Leyvit\n'
        '⚠️ Datos de declaraciones públicas en medios — pendiente validar con dictamen técnico (MIA)',
        transform=ax.transAxes, ha='center', fontsize=8, color='gray')

plt.tight_layout()

# Guardar en /content/visualizations (ruta correcta para Google Colab)
output_dir = '/content/visualizations'
os.makedirs(output_dir, exist_ok=True)
plt.savefig(os.path.join(output_dir, '01_evolucion_arboles_oficiales.png'), dpi=150, bbox_inches='tight')
plt.show()
print('✅ Gráfica guardada en /content/visualizations/')
```
